```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
iris=px.data.iris()
iris
```

```python
#stripplot
#axes level
sns.stripplot(data=tips,x='day',y='total_bill',jitter=False)
```

```python
#using catplot   ----- figure level
sns.catplot(data=tips,x='day',y='total_bill',kind='strip',jitter=False)
```

```python
#jitter=values
#hue
sns.stripplot(data=tips,x='day',y='total_bill',jitter=0.2,hue='sex')
```

```python
#swarmplot
sns.swarmplot(data=tips,x='day',y='total_bill')
```

```python
#hue
sns.swarmplot(data=tips,x='day',y='total_bill',hue='sex')
```

```python
#boxplot
sns.boxplot(data=tips,x='sex',y='total_bill')
```

```python
#using catplot
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='box',hue='sex')
```

```python
#single boxplot---.numerical col
sns.catplot(data=tips,y='total_bill',kind='box')
```

```python
#violinplot(kde+box)
sns.catplot(data=tips,x='day',y='total_bill',kind='violin')
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='violin',hue='sex',split=True)
```

```python
#barplot
sns.catplot(data=tips,x='day',y='total_bill',kind='bar',hue='sex',estimator=np.std)
```

```python
#black color line indiactes-when there are multiple obervations in eact category,
#it also uses bootstrapping to compute a confindences interval arounds the estimates, 
#which is plotted using error bars.

sns.catplot(data=tips,x='day',y='total_bill',kind='bar',hue='sex',errorbar=None)
```

```python
#point plot
sns.catplot(data=tips,x='day',y='total_bill',kind='point',hue='sex',estimator=np.std)
```

```python
sns.catplot(data=tips,x='day',y='total_bill',kind='point',ci=None)
```

```python
#countplot
sns.countplot(data=tips,x='sex',hue='day')
```

```python
#facting using catplot
sns.catplot(data=tips,x='day',y='total_bill',kind='box',col='smoker')
```

```python
#reg plot----hue is not available
sns.regplot(data=tips,x='total_bill',y='tip')
```

```python
#lmplot==hue
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
```

```python
#residplot
sns.residplot(data=tips,x='total_bill',y='tip')
```

```python
#figure level->relplot,displot,catplot,lmplot
sns.catplot(data=tips,x='total_bill',y='tip',col='sex',row='time',kind='violin')
```

```python
g=sns.FacetGrid(data=tips,col='day',row='time')
g.map(sns.violinplot,'sex','total_bill')
```

```python
#plotting pairwise relationship(pairGrid vs pairplot)
```

```python
iris
sns.pairplot(iris,hue='species')
```

```python
#pairgrid
g=sns.PairGrid(data=iris)
g.map(sns.scatterplot)
```

```python
#map_diag---> map_offdiag
g=sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
#jointgrid vs joint plot
sns.jointplot(data=tips,x='total_bill',y='tip',kind='kde',hue='sex')
```

```python
g=sns.JointGrid(data=tips)
g.plot(sns.scatterplot,sns.histplot)
```
