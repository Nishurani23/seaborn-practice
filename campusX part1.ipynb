```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
#scatterplot---axes level
tips=sns.load_dataset('tips')
tips
```

```python
sns.scatterplot(data=tips, x='total_bill',y='tip')
```

```python
#replot----fiqure level
sns.relplot(data=tips, x='total_bill',y='tip',kind ='scatter',hue='sex',style='time',size='size')
```

```python
#hue
#style
#size
sns.scatterplot(data=tips, x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
#line chart
gap= px.data.gapminder()
temp=gap[gap['country']=='India']
temp
```

```python
#axes level functions
sns.lineplot(data=temp, x='year',y='lifeExp')
```

```python
#relplot
sns.relplot(data=temp, x='year',y='lifeExp' , kind ='line')
```

```python
#hue
#style
temp=gap[gap['country'].isin(['India','Pakistan','China'])]
temp
sns.relplot(data=temp, x='year',y='lifeExp' , kind ='line',hue='country')
```

```python
sns.lineplot(data=temp, x='year',y='lifeExp',hue='country')
```

```python
temp=gap[gap['country'].isin(['India','Brazil','Germany'])]
temp
sns.relplot(data=temp, x='year',y='lifeExp' , kind ='line',hue='country',style='continent',size='continent')
```

```python
#facet plot------figure level function with relplot
sns.relplot(data=tips, x='total_bill',y='tip',kind ='scatter',col='sex',row='smoker')
```

```python
#col wrap
sns.relplot(data=gap, x='lifeExp',y='gdpPercap' , kind ='scatter',col='year',col_wrap=5)
```

```python
#potting histogram
sns.histplot(data=tips, x='total_bill')
```

```python
sns.displot(data=tips, x='total_bill', kind='hist')
```

```python
#bins
sns.displot(data=tips, x='total_bill', kind='hist',bins=5)
```

```python
sns.displot(data=tips, x='day', kind='hist')
```

```python
sns.displot(data=tips, x='day', kind='hist',hue='sex',bins=10)
```

```python
#elements-----step
sns.displot(data=tips, x='day', kind='hist',hue='sex',element='step')
```

```python
#heatmap======axes level
gap
column=['lifeExp',	'pop',	'gdpPercap']
        

sns.heatmap(data=gap[column].corr(),annot=True,cmap="Blues")
```

```python
#plot rectangle date as a color-encoded matrix
temp=gap.pivot(index='country',columns='year',values='lifeExp')
sns.heatmap(temp)
```

```python
#annot ,cmap
temp=gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')
sns.heatmap(temp,annot=True,cmap='Blues')
```
