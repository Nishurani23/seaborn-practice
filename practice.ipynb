```python
#Create a Bar Plot Showing Total Confirmed Cases by WHO Region
region_cases = (df.groupby('WHO Region')['Confirmed'] .sum().reset_index())
sns.barplot(data=region_cases,x='WHO Region',y='Confirmed')
```

```python
#Create a bar plot showing total deaths by WHO Region.
region_cases = (df.groupby('WHO Region')['Deaths'] .sum().reset_index())
sns.barplot(data=region_cases,x='WHO Region',y='Deaths')
```

```python
#Create a scatter plot of Confirmed vs Deaths.
sns.scatterplot(data=covid,x='Confirmed',y='Deaths')
#Color the scatter plot using WHO Region (hue).
sns.scatterplot(
    data=df,
    x='Confirmed',
    y='Deaths',
    hue='WHO Region'
)

#Use both hue and size in a scatter plot.
sns.scatterplot(
    data=df,
    x='Confirmed',
    y='Deaths',
    hue='WHO Region',
    size='Recovered',
    alpha=0.7
)

#Compare Confirmed and Deaths using different marker styles.
sns.scatterplot(
    data=df,
    x='Confirmed',
    y='Deaths',
    hue='WHO Region',
    style='WHO Region',  # Different marker for each region
    s=100)
```

```python
#Create a box plot of Confirmed cases.
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='Confirmed')
plt.show()


#Create a box plot of Confirmed cases grouped by WHO Region.
sns.boxplot(
    data=df,
    x='WHO Region',
    y='Confirmed'
)

top10 = df.sort_values(
    by='Confirmed',
    ascending=False
).head(10)

plt.figure(figsize=(12,6))

sns.boxplot(
    data=top10,
    x='Country/Region',
    y='Confirmed'
)
```
