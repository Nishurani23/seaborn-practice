```python
#the ways of style themes are as follow
#white,dark,whitegrid,darkgrid,ticks

sns.set_style("whitegrid")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g", hue='island')

```

```python
#kde
sns.kdeplot(data=penguins,x="body_mass_g",hue="island",fill=True ,palette="husl")
```

```python
#heatmap
penguins.head(1)
column=['bill_length_mm',	'bill_depth_mm',	'flipper_length_mm',	'body_mass_g']

sns.heatmap(data=penguins[column].corr(),annot=True,cmap="Blues")
```

```python
sns.heatmap(data=penguins[column].corr(),annot=True,cmap="Blues",vmin=-0.2,linewidth="2",linecolor="black")
```
